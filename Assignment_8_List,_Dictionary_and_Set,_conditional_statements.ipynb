```python
age_list = [24, 25, 26, 27, 28]
print(age_list)
```

```python
name_list = ["Ammu", "Appu", "Dachu", "Ishi", "Suru"]
print(name_list)
```

```python
name_list.append("Yazhini")
print(name_list)
```

```python
age_list.insert(2, 30)
print(age_list)
```

```python
name_list.remove("Yazhini")
print(name_list)
```

```python
age_list.pop()
print(age_list)
```

```python
age_list.extend([29, 30, 26])
print(age_list)
```

```python
age_list.sort(reverse=True)
print(age_list)
```

```python
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
print(name_list[0])
print(name_list[-1])
print(name_list[2:5])
print(name_list[::-1])
```

```python
student_marks = {
    "Ammu": 85,
    "Appu": 90,
    "Dachu": 76,
    "Ishi": 88,
    "Suru": 95
}

print(student_marks["Ammu"])
student_marks["Janan"] = 80
student_marks["Ammu"] = 82
print(student_marks)
```

```python
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Key-Value pairs:", student_marks.items())
```

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'i'}

print(my_set)
```

```python
my_set[4] = 's'

Explanation: Sets are unordered, so they don't have indexes like lists. Therefore, we cannot use my_set[4] to access or modify an element.
```

```python
set1 = {1, 3, 5, 7, 9}

set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)
```

```python
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
